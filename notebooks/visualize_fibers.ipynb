```python
from pathlib import Path

import numpy as np
import pyvista as pv

from cardiac_electrophysiology.utils import mesh_processing as mp
```

```python
patient_id = "01"
mesh_path = Path(f"../data/patient_{patient_id}/mesh_endo_raw.vtu")

input_mesh = pv.read(mesh_path)
triangular_mesh = mp.convert_unstructured_to_polydata_mesh(input_mesh)
mesh_point_data = triangular_mesh.cell_data_to_point_data()
mesh_point_data.set_active_vectors("fibers")
fiber_glyphs = mesh_point_data.glyph(geom=pv.Arrow())
```

```python
plotter = pv.Plotter(window_size=[900, 900])
plotter.add_mesh(fiber_glyphs)
plotter.add_mesh(triangular_mesh, opacity=0.5)
plotter.show()
```
